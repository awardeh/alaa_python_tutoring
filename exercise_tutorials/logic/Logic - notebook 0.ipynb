```python
import matplotlib.pyplot as plt
import random
from math import pi

Number_of_particles = 0
radius = 0.5
L = 11
max_particles = (L**2) // (pi * (radius**2))

max_attempts = 10


greatest_x = []
greatest_y = []

while max_attempts > 0 and Number_of_particles <= max_particles:
    x = random.uniform(0, 12)
    y = random.uniform(0, 12)
    listx = []
    listy = []
    max_random = random.uniform(0, max_particles)
    listx.append(x)
    listy.append(y)
    last_idx = 0
    particles = 0
    while particles < max_random:
        x = random.uniform(0, 12)
        y = random.uniform(0, 12)

        if ((x - listx[last_idx]) ** 2) / (y - listy[last_idx]) ** 2 >= 0.5:
            listx.append(x)
            listy.append(y)
            particles += 1
            last_idx += 1
    # print(len(listx))
    if len(listx) >= len(greatest_x):
        greatest_x = listx
        greatest_y = listy
    Number_of_particles = particles
    max_attempts -= 1


plt.scatter(greatest_x, greatest_y, s=100)
plt.show()

print(len(greatest_x))

# listx = [1, 2]
# listy = [1, 2, 3]
```
